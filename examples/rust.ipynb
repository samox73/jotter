# Rust in jOtter

Needs the [evcxr](https://github.com/evcxr/evcxr) kernel. The first cell takes a moment: every cell is compiled. Variables and functions stay defined across cells.

```rust
let xs: Vec<f64> = (1..=10).map(|n| n as f64).collect();
let mean: f64 = xs.iter().sum::<f64>() / xs.len() as f64;
println!("mean of {} numbers: {mean}", xs.len());
```

```rust
fn variance(v: &[f64]) -> f64 {
    let m = v.iter().sum::<f64>() / v.len() as f64;
    v.iter().map(|x| (x - m).powi(2)).sum::<f64>() / v.len() as f64
}

variance(&xs)
```

```rust
#[derive(Debug)]
struct Stats {
    mean: f64,
    sd: f64,
}

let stats: Stats = Stats { mean, sd: variance(&xs).sqrt() };
stats
```

An expression without a trailing `;` is the cell's value, shown with `Debug`:

```rust
xs.iter().filter(|x| **x > stats.mean).collect::<Vec<_>>()
```
