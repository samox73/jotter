# R in jOtter

Needs the [IRkernel](https://irkernel.github.io) kernel. Everything here is base R.

```r
set.seed(1)
x <- seq(0, 10, length.out = 50)
y <- 2.5 * x + 1 + rnorm(50, sd = 2)
d <- data.frame(x, y)
head(d)
```

```r
fit <- lm(y ~ x, data = d)
coef(fit)
```

```r
plot(d, main = "a noisy line", pch = 19, col = "steelblue")
abline(fit, col = "firebrick", lwd = 2)
```

```r
rms <- function(v) sqrt(mean(v^2))
cat("residual rms:", round(rms(resid(fit)), 3), "\n")
```

Tables print as plain text; a data frame is just the value of a cell:
