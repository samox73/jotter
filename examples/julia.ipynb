# Julia in jOtter

Needs the [IJulia](https://github.com/JuliaLang/IJulia.jl) kernel. If your Julia isn't 1.12, jOtter asks which installed Julia kernel to start instead.

```julia
using LinearAlgebra

# harmonic chain: nearest-neighbour springs between N masses
N = 6
K = SymTridiagonal(fill(2.0, N), fill(-1.0, N - 1))
```

```julia
ω = sqrt.(eigvals(K))   # normal-mode frequencies
```

```julia
struct Mode
    index::Int
    frequency::Float64
end

modes = [Mode(i, f) for (i, f) in enumerate(ω)]
for m in modes
    println("mode $(m.index): ω = $(round(m.frequency, digits = 3))")
end
```

The exact dispersion is $\omega_k = 2\sin\frac{k\pi}{2(N+1)}$, so the difference should vanish:

```julia
exact = [2sin(k * π / (2(N + 1))) for k in 1:N]
maximum(abs.(ω .- exact))
```
