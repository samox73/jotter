# Damped oscillator

The displacement of a damped harmonic oscillator is

$$x(t) = A\,e^{-\gamma t}\cos(\omega t + \varphi)$$

with damping rate $\gamma$ and angular frequency $\omega$.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('dark_background')
plt.rcParams.update({'figure.facecolor': '#2b303b', 'axes.facecolor': '#2b303b', 'figure.figsize': (6.5, 3), 'figure.dpi': 200})
```

```python
t = np.linspace(0, 10, 500)
gamma, omega = 0.3, 4.0
x = np.exp(-gamma * t) * np.cos(omega * t)
print(f"{len(t)} samples, peak {x.max():.2f}")
```

```python
plt.plot(t, x, label="$x(t)$")
plt.plot(t, np.exp(-gamma * t), "--", alpha=0.6, label="envelope")
plt.xlabel("$t$")
plt.legend()
plt.show()
```
