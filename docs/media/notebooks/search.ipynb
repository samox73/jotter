```python
import numpy as np
import pandas as pd
```

```python
runs = pd.read_csv("runs.csv")
```

```python
runs = runs.dropna()
runs = runs[runs.energy < 10]
```

```python
def model(t, a, tau):
    return a * np.exp(-t / tau)
```

```python
from scipy.optimize import curve_fit
```

```python
params, cov = curve_fit(model, runs.t, runs.energy)
```

```python
errors = np.sqrt(np.diag(cov))
```

```python
print(f"tau = {params[1]:.3f} ± {errors[1]:.3f}")
```

```python
residuals = runs.energy - model(runs.t, *params)
```

```python
chi2 = (residuals**2).sum() / (len(runs) - 2)
```

```python
runs.assign(fit=model(runs.t, *params)).to_csv("fit.csv")
```
