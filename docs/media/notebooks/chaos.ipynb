```python
# Setup: integrate the Lorenz system and pre-render 48 frames of it
import time, sympy as sp
from scipy.integrate import solve_ivp
from IPython.display import Image, display
from tqdm import tqdm
from chaos import frame
sp.init_printing(use_latex="mathjax")
σ, ρ, β = 10, 28, sp.Rational(8, 3)
x, y, z = sp.symbols("x y z")
lorenz = sp.lambdify([x, y, z], [σ*(y-x), x*(ρ-z)-y, x*y-β*z])
path = solve_ivp(lambda t, s: lorenz(*s), (0, 40), [1, 1, 1], max_step=0.01).y
frames = [frame(path, f, 48) for f in tqdm(range(48), colour="#ffaa00")]
```

## Chaos, live in your terminal 🦋

$$\dot x = \sigma(y-x),\quad \dot y = x(\rho-z)-y,\quad \dot z = xy-\beta z$$

$$z_{n+1} = z_n^2 + c$$

```python
# 1. The Lorenz system's fixed points, found with SymPy
sp.Matrix(sp.solve([σ*(y-x), x*(ρ-z)-y, x*y-β*z], [x, y, z]))
```

```python
# 2. The attractor, and a Mandelbrot dive: the frames played inline
out = display(Image(frames[0]), display_id=True)
for png in frames:
    out.update(Image(png)); time.sleep(1/16)
```
