```python
# 3. A line plot: inside the terminal, never as a separate window
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 4 * np.pi, 400)
fig, ax = plt.subplots(figsize=(6, 3))
for k in range(1, 4):
    ax.plot(x, np.sin(k * x) / k, label=f"$\\sin({k}x)/{k}$")
ax.legend(); ax.set_title("sines")
fig.tight_layout(); plt.show()
```

```python
# 4. A colour image: a smooth, complete viridis gradient, nothing garbled
c = np.linspace(-2.2, 0.8, 240)[None, :] + 1j * np.linspace(-1.2, 1.2, 160)[:, None]
z = np.zeros_like(c); n = np.zeros(c.shape, int)
for i in range(40):
    m = abs(z) < 2
    z[m] = z[m] ** 2 + c[m]; n[m] = i
fig, ax = plt.subplots(figsize=(6, 4))
ax.imshow(n, cmap="viridis"); ax.set_axis_off(); plt.show()
```

```python
# 5. SymPy returns LaTeX: drawn as an equation
import sympy as sp
x = sp.symbols("x")
sp.Integral(sp.sin(x) ** 2, x).doit()
```
